# S&P 500 Price Chart
Fetches historical S&P 500 price data from Yahoo Finance and plots it.

```python
import yfinance as yf
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
```

```python
sp500 = yf.download('^GSPC', period='1y', interval='1d')
print(f'Fetched {len(sp500)} trading days of S&P 500 data')
sp500.tail()
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8), gridspec_kw={'height_ratios': [3, 1]}, sharex=True)
fig.suptitle('S&P 500 Index — Past 12 Months', fontsize=16, fontweight='bold')

close = sp500['Close'].squeeze()
ma50 = close.rolling(window=50).mean()
ma200 = close.rolling(window=200).mean()

ax1.plot(close.index, close.values, color='#1f77b4', linewidth=1.2, label='Close')
ax1.plot(ma50.index, ma50.values, color='#ff7f0e', linewidth=1, linestyle='--', label='50-day MA')
ax1.plot(ma200.index, ma200.values, color='#2ca02c', linewidth=1, linestyle='--', label='200-day MA')
ax1.fill_between(close.index, close.values.flatten(), alpha=0.08, color='#1f77b4')
ax1.set_ylabel('Price (USD)', fontsize=12)
ax1.legend(loc='upper left', fontsize=10)
ax1.grid(True, alpha=0.3)

volume = sp500['Volume'].squeeze()
colors = ['#2ca02c' if close.values.flatten()[i] >= close.values.flatten()[max(0, i-1)] else '#d62728' for i in range(len(close))]
ax2.bar(volume.index, volume.values, color=colors, alpha=0.6, width=1)
ax2.set_ylabel('Volume', fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax2.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
fig.autofmt_xdate()

plt.tight_layout()
plt.savefig('sp500_chart.png', dpi=150, bbox_inches='tight')
plt.show()
print('Chart saved to sp500_chart.png')
```

```python
latest = close.iloc[-1]
prev = close.iloc[-2]
change_pct = (latest - prev) / prev * 100
high_52w = close.max()
low_52w = close.min()

print(f'Latest Close:  ${latest:,.2f}')
print(f'Daily Change:  {change_pct:+.2f}%')
print(f'52-Week High:  ${high_52w:,.2f}')
print(f'52-Week Low:   ${low_52w:,.2f}')
```
